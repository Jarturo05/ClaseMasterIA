* MCAR - Falta Completamente al azar: Cuando el nulo no tiene relacion con ninguna variable o el valor que falta
* MAR - Relacionado con otra variables: Tiene un valor nulo pero tiene un relacion con una variable
* MNAR - Relacionado con el valor que falta: El nulo esta directamente relacionado con el valor que deberia estar ahi

Imputar - Un valor faltante por un valor estimado

* Media: Numerica, distribucion simetrica, sin outlier. La suma de todos los valores dividio en la cantidad de los valores
* Mediana: Numerica con outliers o distribucion sesgada. El Valor central // La mejor alternativa para rellenar los datos nulos que tenga valores exagerados
* Moda: Variable categorica. El valor que aparece con mayor frecuencia o el mas repetido
* Por grupo: El nulo depende de otra variable. 

```python
#Imputacion 1 Salario ---> Mediana

mediana = df["salario_cop"].median()
print(f"Mediana de salario: ${mediana:,.0f}")

df["salario_cop"] = df["salario_cop"].fillna(mediana)
```

```python
df.isnull().sum()
```

```python
#Imputacion 2 Nivel de estres ---> Media

media = df["nivel_estres"].mean()
print(f"La media del nivel de estres es: {media:.4f}")

df['nivel_estres'] = df['nivel_estres'].fillna(media)
```

```python
#Imputacion 3 Satisfaccion Laboral ---> Media
media_satisfaccion_laboral = df['satisfaccion_laboral'].mean()
print(f"La media para satisfaccion laboral: {media_satisfaccion_laboral}")

df['satisfaccion_laboral'] = df['satisfaccion_laboral'].fillna(media_satisfaccion_laboral)
```

```python
#Imputacion 4 ciudad con moda 

ciudad_moda = df['ciudad'].mode()[0]
print(f"Moda de la ciudad: {ciudad_moda}")

df['ciudad'] = df['ciudad'].fillna(ciudad_moda)
```

```python
df.isnull().sum()
```

## Rango Intercuartil
$$
IQR = Q3 - Q1
$$
$$
Limite\;superior = Q1 - (1.5*IQR)
$$
$$
Limite\;inferior = Q3 + (1.5*IQR)
$$


```python
# Cuartiles

Q1 = df ['horas_semana'].quantile(0.25)
Q3 = df ['horas_semana'].quantile(0.75)

IQR = Q3 - Q1

print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
print(f"IQR: {IQR}")
```

```python
# Limite Inferior y Superior

L_I = Q1 - (1.5 * IQR)
print(f"El limite Inferior es: {L_I:.2f} horas")

L_S = Q3 + (1.5 * IQR)
print(f"El limite Inferior es: {L_S:.2f} horas")
```

```python
outliers_iqr = df[
(df['horas_semana'] < L_I) |
(df['horas_semana'] > L_S)
]
print(f"Outliers detectados en horas_semana: {len(outliers_iqr)}")
print(outliers_iqr[['id_trabajador', 'cargo', 'nivel', 'horas_semana']])
```
