## Listas y Diccionarios

> Listas:
Colección ordenada, implementada como un array dinámico, su acceso lo tiene por indices.


```python
# Lista de estudiantes

estudiantes = ["Graciela", "Andres", "Arturo"]

print(estudiantes[1])
print(estudiantes[-1]) # Acceder al ultimo dato de la lista
print(len(estudiantes))

# Agregando un estudiante
estudiantes.append("Camila")
print(estudiantes)
```

# Condicionales y bucles

> Condicionales -> if/elif/else, es una estructura de control del flujo


```python
nombre = "Ana"
promedio = 8.7

if promedio >= 7:
    print(f"Felicitaciones {nombre}")
elif promedio >=5:
    print(f"{nombre} Puedes mejorar")
else:
    print(f"Necesitas mejorar")
```

```python
learners = [
    {"nombre": "Pepita", "promedio": 8.7}
    {"nombre": "Pepito", "promedio": 7.5}
    {"nombre": "Andres", "promedio": 9.1}
]

for learner in learners:
    if learner["promedio"] >= 7:
        print(f"Felicitaciones {nombre}")
    elif learner["promedio"] >=5:
        print(f"{nombre} Puedes mejorar")
    else:
        print(f"Necesitas mejorar")
```
